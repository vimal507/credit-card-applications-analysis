```python
plt.figure(figsize=(10,5))
sns.countplot(data=credit,x="owner",hue="card")
plt.title("Home ownership vs card approval")
plt.xlabel("Home ownership")
plt.ylabel("Number of applicants")
plt.legend(title="Card approval")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(credit["expenditure"],bins=30,kde=True)
plt.title("Customer expenditure distribution")
plt.xlabel("expenditure")
plt.ylabel("frequency")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=credit,x="card")
plt.title("credit card approval distribution")
plt.xlabel("card approval")
plt.ylabel("Number of applications")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=credit,x="card",y="income")
plt.title("income vs credit card approval")
plt.xlabel("card approval")
plt.ylabel("income")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data =credit ,x="income",y="expenditure",hue="card")
plt.title("income vs expenditure")
plt.xlabel("income")
plt.ylabel("expenditure")
plt.show()
```

## insights

1. credit history - Applicants with zero major derogatory reports have an approval rate of approximately 86.32%. Applicants with one or more reports have an approval rate of approximately 41.70%.


2.Home Ownership - 
Homeowners have an approval rate of approximately 84.51%, compared with 72.09% for non-homeowners.
